```python
import pandas as pd
import os

data_path = "../data"

files = {
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "customers": "olist_customers_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "products": "olist_products_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "category_translation": "product_category_name_translation.csv"
}

data = {}

for name, file in files.items():
    path = os.path.join(data_path, file)
    data[name] = pd.read_csv(path)
    print(f"{name}: {data[name].shape}")
```

```python
for name, df in data.items():
    print("\n" + "="*50)
    print(name.upper())
    print("="*50)
    print(df.columns.tolist())
```

```python
for name, df in data.items():
    print("\n" + "="*50)
    print(name.upper())
    print("="*50)

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(missing)
```

```python
orders = data["orders"].copy()

print("Missing values before handling:")
print(orders.isnull().sum())

print("\nMissing values are retained because they represent")
print("orders where the corresponding process stage did not occur.")
```

```python
data["orders"] = orders

print("Orders dataset updated successfully.")
```

```python
for name, df in data.items():
    print(name, "→ duplicate rows:", df.duplicated().sum())
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    data["orders"][col] = pd.to_datetime(data["orders"][col])

print(data["orders"][date_columns].dtypes)
```

```python
print("Purchase date range:")
print(data["orders"]["order_purchase_timestamp"].min())
print(data["orders"]["order_purchase_timestamp"].max())
```

```python
products = data["products"]

print("Number of unique product categories:")
print(products["product_category_name"].nunique())

print("\nSample categories:")
print(products["product_category_name"].dropna().unique()[:30])
```

```python
products = data["products"].copy()
translation = data["category_translation"].copy()

products = products.merge(
    translation,
    on="product_category_name",
    how="left"
)

print(products[[
    "product_category_name",
    "product_category_name_english"
]].head(10))
```

```python
print("Categories before translation:",
      products["product_category_name"].nunique())

print("English categories available:",
      products["product_category_name_english"].nunique())

print("Missing translations:",
      products["product_category_name_english"].isnull().sum())
```

```python
products["product_category_final"] = products[
    "product_category_name_english"
].fillna(
    products["product_category_name"]
)

print("Missing final categories:",
      products["product_category_final"].isnull().sum())

print("\nSample final categories:")
print(products["product_category_final"].head(10))
```

```python
data["products"] = products

print("Products dataset updated successfully.")
```

```python
order_items = data["order_items"]

print("PRICE")
print(order_items["price"].describe())

print("\nFREIGHT VALUE")
print(order_items["freight_value"].describe())
```

```python
for col in ["price", "freight_value"]:
    Q1 = order_items[col].quantile(0.25)
    Q3 = order_items[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = order_items[
        (order_items[col] < lower) |
        (order_items[col] > upper)
    ]

    print("\n", col)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Number of outliers:", len(outliers))
```

```python
for name, df in data.items():
    print("\n" + "="*40)
    print(name.upper())
    print("="*40)
    print(df.dtypes)
```

```python
data["order_items"]["shipping_limit_date"] = pd.to_datetime(
    data["order_items"]["shipping_limit_date"]
)

print(data["order_items"]["shipping_limit_date"].dtype)
```

```python
checks = {
    "orders → customers":
        data["orders"]["customer_id"].isin(
            data["customers"]["customer_id"]
        ).all(),

    "order_items → orders":
        data["order_items"]["order_id"].isin(
            data["orders"]["order_id"]
        ).all(),

    "order_items → products":
        data["order_items"]["product_id"].isin(
            data["products"]["product_id"]
        ).all(),

    "order_items → sellers":
        data["order_items"]["seller_id"].isin(
            data["sellers"]["seller_id"]
        ).all(),

    "payments → orders":
        data["payments"]["order_id"].isin(
            data["orders"]["order_id"]
        ).all(),

    "reviews → orders":
        data["reviews"]["order_id"].isin(
            data["orders"]["order_id"]
        ).all()
}

for relationship, result in checks.items():
    print(relationship, "→", result)
```

```python
# Prepare order-level payments
payments_summary = (
    data["payments"]
    .groupby("order_id")
    .agg(
        total_payment=("payment_value", "sum"),
        payment_count=("payment_sequential", "count"),
        payment_types=("payment_type", "nunique")
    )
    .reset_index()
)

# Prepare order-level reviews
reviews_summary = (
    data["reviews"]
    .groupby("order_id")
    .agg(
        review_score=("review_score", "mean"),
        review_count=("review_id", "count")
    )
    .reset_index()

)

# Prepare order-item information
items_summary = (
    data["order_items"]
    .groupby("order_id")
    .agg(
        item_count=("order_item_id", "count"),
        total_price=("price", "sum"),
        total_freight=("freight_value", "sum"),
        seller_count=("seller_id", "nunique"),
        product_count=("product_id", "nunique")
    )
    .reset_index()
)

print("Payments summary:", payments_summary.shape)
print("Reviews summary:", reviews_summary.shape)
print("Items summary:", items_summary.shape)
```

```python
# Start with orders
analytical_data = data["orders"].copy()

# Add customer information
analytical_data = analytical_data.merge(
    data["customers"],
    on="customer_id",
    how="left"
)

# Add item-level summary
analytical_data = analytical_data.merge(
    items_summary,
    on="order_id",
    how="left"
)

# Add payment summary
analytical_data = analytical_data.merge(
    payments_summary,
    on="order_id",
    how="left"
)

# Add review summary
analytical_data = analytical_data.merge(
    reviews_summary,
    on="order_id",
    how="left"
)

print("Final analytical dataset shape:", analytical_data.shape)
print("\nColumns:")
print(analytical_data.columns.tolist())

print("\nDuplicate order IDs:",
      analytical_data["order_id"].duplicated().sum())
```

```python
import sqlite3

# Create SQLite database
conn = sqlite3.connect("../olist_analytics.db")

# Load analytical dataset into SQL table
analytical_data.to_sql(
    "orders_analytics",
    conn,
    if_exists="replace",
    index=False
)

print("SQL table created successfully.")
print("Rows loaded:", len(analytical_data))
```
