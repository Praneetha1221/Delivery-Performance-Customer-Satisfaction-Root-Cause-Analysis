```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(".env")

api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

findings = """
Olist E-commerce Customer Experience Analysis — Initial Findings

1. Total revenue: 13,591,643.70
2. Average order value: 137.75
3. Average review score: 4.09 / 5
4. Late delivery rate: 8.11%
5. Average actual delivery time: 12.56 days
6. Average estimated delivery time: 23.74 days

Delivery vs Review Score:
- More than 7 days early: 4.32
- 1–7 days early: 4.20
- On time: 3.18
- 1–7 days late: 1.70
- More than 7 days late: 2.02

Order Value vs Review Score:
- Under 50: 4.17
- 50–100: 4.12
- 100–250: 4.08
- 250–500: 3.97
- 500+: 3.94

Freight Ratio vs Review Score:
- 0–5%: 4.09
- 5–10%: 4.15
- 10–20%: 4.13
- 20–50%: 4.10
- 50%+: 4.06

Complaint themes among negative reviews:
- Missing product: 25.10%
- Late delivery: 21.34%
- Product quality: 10.47%
- Seller issue: 5.08%
- Product damage: 2.82%
- Wrong product: 2.36%
- Shipping: 2.08%
- Cancellation / Refund: 1.31%
- Packaging: 1.26%
- Payment: 0.87%
- Other / unclear: 27.32%

Category findings:
- Some product categories have substantially lower review scores than others.
- Office furniture was among the lowest-rated categories.

Seller findings:
- Seller performance varies substantially.
- Some sellers have consistently low review scores and poor customer experience indicators.

Geographic findings:
- Review scores vary across Brazilian states.
- São Paulo has the highest order volume and revenue.
"""

prompt = f"""
You are a senior e-commerce data analyst.

Based ONLY on the initial findings below, generate possible
business hypotheses that could explain poor customer satisfaction.

Important:
- Do NOT claim that any hypothesis is proven.
- Do NOT treat correlation as causation.
- Each hypothesis must be testable using data.
- Explain what data or analysis could be used to test each hypothesis.

Generate hypotheses covering:
1. Delivery
2. Sellers
3. Product categories
4. Freight costs
5. Geography
6. Any additional important hypothesis you identify

For each hypothesis provide:
- Hypothesis
- Why the finding suggests it
- What data/analysis should test it

Initial findings:
{findings}
"""

response = client.responses.create(
    model="gpt-5.6-luna",
    input=prompt
)

print(response.output_text)
```
