# Final Business Insights

## Finding 1 — Late Delivery and Customer Satisfaction

**FINDING:**  
Late deliveries are strongly associated with lower customer ratings.

**EVIDENCE:**  
Late orders had an average rating of 2.57, compared with 4.29 for on-time/early orders. The difference was 1.73 rating points, with a statistically significant result (p < 0.001).

**BUSINESS IMPACT:**  
Delivery reliability is strongly associated with customer satisfaction.

**RECOMMENDATION:**  
Prioritize late-delivery reduction through seller SLA monitoring, delivery tracking, and early intervention.

---

## Finding 2 — Severe Delivery Delays

**FINDING:**  
Severe delivery delays are associated with highly negative customer experiences.

**EVIDENCE:**  
Orders delivered more than 7 days late had an average rating of 1.73, and 87.46% of their reviews were negative.

**BUSINESS IMPACT:**  
Severely delayed orders represent a high-risk customer experience.

**RECOMMENDATION:**  
Create an escalation process for orders approaching or exceeding the estimated delivery date.

---

## Finding 3 — Seller Performance Opportunities

**FINDING:**  
Some high-revenue sellers have weaker customer-experience performance.

**EVIDENCE:**  
871 sellers were identified as Priority Problem sellers, representing approximately $8.68M in revenue. Their average rating was 3.75 and late-delivery rate was 11.69%.

**BUSINESS IMPACT:**  
Improving high-revenue underperforming sellers could have a meaningful business impact.

**RECOMMENDATION:**  
Prioritize these sellers for SLA reviews, quality checks, and operational support.

---

## Finding 4 — High-Revenue, Low-Rating Categories

**FINDING:**  
Several high-revenue product categories also show below-average customer ratings.

**EVIDENCE:**  
Examples include Watches & Gifts (~$1.21M, rating 4.02), Bed/Bath/Table (~$1.05M, rating 3.90), Computers/Accessories (~$920K, rating 3.93), and Furniture/Decor (~$736K, rating 3.90).

**BUSINESS IMPACT:**  
Poor customer experiences in high-revenue categories can affect a significant portion of business activity.

**RECOMMENDATION:**  
Prioritize quality, packaging, seller selection, and delivery improvements in these categories.

---

## Finding 5 — Major Customer Complaint Themes

**FINDING:**  
Negative customer reviews are concentrated around a few recurring complaint themes.

**EVIDENCE:**  
Among 14,445 negative reviews with text:
- Missing product: 25.10%
- Late delivery: 21.34%
- Product quality: 10.47%
- Seller issue: 5.08%

These four identifiable themes represent approximately 62% of negative reviews.

**BUSINESS IMPACT:**  
Customer dissatisfaction is concentrated around fulfillment, delivery, quality, and seller-related issues.

**RECOMMENDATION:**  
Focus improvement efforts on order completeness, delivery reliability, product quality, and seller management.

---

## Finding 6 — Geographic Experience Differences

**FINDING:**  
Customer experience varies significantly across regions.

**EVIDENCE:**  
Among states with sufficient order volume, some of the lowest average ratings were AL (3.75), MA (3.76), SE (3.81), CE (3.85), and RJ (3.87).

**BUSINESS IMPACT:**  
Some regions may require closer operational monitoring.

**RECOMMENDATION:**  
Investigate regional delivery networks, seller distribution, freight costs, and service coverage in low-performing regions.

---

## Finding 7 — Order Value and Customer Ratings

**FINDING:**  
Higher-value orders show slightly lower average customer ratings.

**EVIDENCE:**  
Average rating decreased from 4.17 for orders below $50 to 3.94 for orders above $500.

**BUSINESS IMPACT:**  
Higher-value purchases may involve greater customer expectations.

**RECOMMENDATION:**  
Provide stronger delivery visibility, quality assurance, and proactive communication for high-value orders.

---

# Overall Business Conclusion

Customer experience is most strongly associated with delivery reliability, seller performance, product quality, and order fulfillment.

The highest-priority opportunities are reducing severe delivery delays, improving high-revenue underperforming sellers and categories, and addressing recurring complaints such as missing products and product-quality issues.
