# Business Recommendations

## Recommendation 1 — Reduce Late Deliveries

**ACTION:**  
Prioritize sellers and delivery routes with consistently high late-delivery rates, especially those handling a large number of orders.

**WHY:**  
Late deliveries are strongly associated with lower customer ratings. Orders delivered late had an average rating of 2.57 compared with 4.29 for on-time/early orders.

**EXPECTED IMPACT:**  
Lower late-delivery rates, improved customer satisfaction, and fewer negative reviews.

**HOW TO MEASURE:**  
Track late-delivery rate, average delivery delay, and average review score before and after implementing delivery improvements.

---

## Recommendation 2 — Escalate Severely Delayed Orders

**ACTION:**  
Create an early-warning and escalation process for orders approaching or exceeding their estimated delivery date.

**WHY:**  
Orders delivered more than 7 days late had an average rating of 1.73, with 87.46% of reviews classified as negative.

**EXPECTED IMPACT:**  
Reduce severe delivery failures and prevent highly dissatisfied customer experiences.

**HOW TO MEASURE:**  
Monitor the percentage of orders delivered more than 7 days late and the negative-review rate for delayed orders.

---

## Recommendation 3 — Prioritize High-Impact Underperforming Sellers

**ACTION:**  
Prioritize sellers with high revenue or order volume and consistently weak ratings or delivery performance for SLA reviews, quality checks, and operational support.

**WHY:**  
Underperforming high-impact sellers affect a larger number of customers and can contribute significantly to poor customer experience.

**EXPECTED IMPACT:**  
Improved seller reliability, higher customer ratings, and reduced negative reviews.

**HOW TO MEASURE:**  
Track seller-level average rating, late-delivery rate, negative-review rate, and revenue before and after intervention.

---

## Recommendation 4 — Investigate High-Revenue, Low-Rating Categories

**ACTION:**  
Investigate high-revenue categories with below-average ratings and high negative-review rates, focusing on product quality, packaging, seller selection, and delivery issues.

**WHY:**  
Categories such as watches/gifts, bed/bath/table, computers/accessories, and furniture/decor generate substantial revenue while showing weaker customer ratings.

**EXPECTED IMPACT:**  
Protect revenue from high-value categories while improving customer satisfaction.

**HOW TO MEASURE:**  
Monitor category-level revenue, average rating, negative-review rate, and late-delivery rate over time.

---

## Recommendation 5 — Reduce Missing-Product Complaints

**ACTION:**  
Investigate recurring missing-product complaints by reviewing order fulfillment, inventory handling, packing, and shipment handoff processes.

**WHY:**  
Missing products were the largest identifiable complaint theme, representing 25.10% of the negative reviews analyzed through the complaint taxonomy.

**EXPECTED IMPACT:**  
Fewer incomplete orders, fewer refunds or complaints, and improved customer trust.

**HOW TO MEASURE:**  
Track missing-product complaint rate, refund/cancellation rate, and negative-review rate related to fulfillment issues.

---

## Recommendation 6 — Improve Product Quality in High-Complaint Categories

**ACTION:**  
Identify categories with high product-quality complaints and review seller quality, product specifications, packaging, and return patterns.

**WHY:**  
Product quality was one of the major recurring themes among negative customer reviews.

**EXPECTED IMPACT:**  
Reduce quality-related dissatisfaction, returns, and negative reviews.

**HOW TO MEASURE:**  
Track product-quality complaint rate, return rate, average rating, and negative-review rate by category.

---

## Recommendation 7 — Monitor Regional Experience Hotspots

**ACTION:**  
Prioritize regions with consistently lower ratings, higher freight costs, or weaker delivery performance for deeper operational investigation.

**WHY:**  
Customer experience differs significantly across regions, with some states showing substantially lower average ratings.

**EXPECTED IMPACT:**  
More consistent customer experience across regions and identification of location-specific operational problems.

**HOW TO MEASURE:**  
Track regional average rating, late-delivery rate, average delivery delay, freight ratio, and negative-review rate.

---

# Recommendation Summary

The highest-priority actions are:

1. Reduce late and severely delayed deliveries.
2. Improve high-impact underperforming sellers.
3. Investigate high-revenue, low-rating categories.
4. Reduce missing-product and quality-related complaints.
5. Monitor regional customer-experience hotspots.

These recommendations connect directly to the analysis and can be measured using operational and customer-experience KPIs.
